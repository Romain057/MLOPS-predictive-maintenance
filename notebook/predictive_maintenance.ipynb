```python
import pandas as pd 
```

```python
df = pd.read_csv("../data/raw/data.csv")
df.head(10)
```

```python
df.describe()
```

```python
df.info()
```

```python
print(df["Machine failure"].value_counts())
panne = (df["Machine failure"] == 1).sum()
print(panne)
fonctionnel = (df["Machine failure"] == 0).sum()
print(fonctionnel)
taux_panne = (panne/fonctionnel) * 100
print(f"Le taux de panne est de {taux_panne:.2f} %")
```

Alternative

```python
counts = df["Machine failure"].value_counts()

panne = counts.get(1, 0)
fonctionnel = counts.get(0, 0)
total = len(df)

taux_panne = panne / total * 100

print(f"Machines en panne : {panne}")
print(f"Machines fonctionnelles : {fonctionnel}")
print(f"Taux de panne global : {taux_panne:.2f} %")
```
